# CUTLOOP — Phase 2: Data Cleaning & Preprocessing
## Fast-Fashion Garment Dataset & TexSortML Preprocessing
This notebook demonstrates the preprocessing pipeline: handling missing values, extracting fiber compositions, and preparing image splits.


```python
import os
import pandas as pd
import numpy as np

base_dir = os.path.abspath('..')
cleaned_csv = os.path.join(base_dir, 'datasets', 'processed', 'garments_cleaned.csv.gz')
manifest_csv = os.path.join(base_dir, 'datasets', 'processed', 'image_splits_manifest.csv')

df_garments = pd.read_csv(cleaned_csv)
print('Garments dataset shape:', df_garments.shape)
df_garments.head(3)
```

### 1. Composition & Blend Statistics
Evaluate single-fiber mono-material vs blended fabric distributions.


```python
print('Mono-material (>=95% single fiber):', df_garments['is_mono_material'].value_counts(normalize=True).round(3))
print('\nBlended materials (>1 fiber):', df_garments['is_blend'].value_counts(normalize=True).round(3))
print('\nDominant Fiber Distribution:\n', df_garments['dominant_fiber'].value_counts().head(8))
```

### 2. TexSortML Image Splits Manifest
Inspect the stratified 70/15/15 train/val/test split.


```python
df_images = pd.read_csv(manifest_csv)
print('Total image records:', len(df_images))
pd.crosstab(df_images['category'], df_images['split'], margins=True)
```
