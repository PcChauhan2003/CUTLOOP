# CUTLOOP Data Science — Notebook 01: Dataset Inspection
**Goal:** Inspect the Fast-Fashion Garment Dataset and the TexSortML Image Dataset to understand data volume, schemas, distributions, and quality limitations.

```python
import os
import json
import zipfile
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter

# Configure visual styling
sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.figsize'] = (10, 5)
print('Imports successful.')
```

## 1. Inspect Fast-Fashion Garment JSONL Dataset
Loading and examining schema for `6_JSONL_component_normalized_public.jsonl`.

```python
jsonl_path = os.path.abspath(r'../../../6_JSONL_component_normalized_public.jsonl')
print(f'Checking: {jsonl_path}')
records = []
with open(jsonl_path, 'r', encoding='utf-8') as f:
    for i in range(5000):  # Sample first 5000 for interactive exploration
        records.append(json.loads(f.readline()))
df_sample = pd.DataFrame(records)
display(df_sample[['product_name', 'brand', 'parent_category', 'detail_category', 'raw_material_text_norm']].head())
```

## 2. Inspect TexSortML Image Dataset
Inspecting the 4,504 JPEG images across stages and categories in `TexSortML-Pictures.zip`.

```python
zip_path = os.path.abspath(r'../../../TexSortML-Pictures.zip')
print(f'Checking: {zip_path}')
with zipfile.ZipFile(zip_path, 'r') as zf:
    categories = Counter(f.split('/')[2] for f in zf.namelist() if f.endswith('.jpg') and len(f.split('/')) >= 3)
plt.bar(categories.keys(), categories.values(), color='#1B4D3E')
plt.xticks(rotation=45, ha='right')
plt.title('TexSortML Image Category Distribution')
plt.ylabel('Image Count')
plt.tight_layout()
plt.show()
```
