## Multimodal Feature Pipelines for Circular Fabric & Garment Modeling
This notebook demonstrates text TF-IDF vectorization, categorical one-hot encoding, circularity indicator scaling, and persistent sparse feature matrix creation.


```python
import os
import joblib
import numpy as np
from scipy.sparse import load_npz

base_dir = os.path.abspath('..')
pipeline_path = os.path.join(base_dir, 'saved_models', 'feature_pipeline.joblib')
X_train = load_npz(os.path.join(base_dir, 'datasets', 'processed', 'X_train_features.npz'))
X_test = load_npz(os.path.join(base_dir, 'datasets', 'processed', 'X_test_features.npz'))
y_train = np.load(os.path.join(base_dir, 'datasets', 'processed', 'y_train.npy'))
y_test = np.load(os.path.join(base_dir, 'datasets', 'processed', 'y_test.npy'))

pipe = joblib.load(pipeline_path)
print('Feature dimensions: Train', X_train.shape, '| Test', X_test.shape)
print('Target classes:', pipe['target_classes'])
```

### 1. TF-IDF Vocabulary Inspection


```python
tfidf = pipe['tfidf']
feature_names = tfidf.get_feature_names_out()
print('Top 20 TF-IDF features:', feature_names[:20])
```
