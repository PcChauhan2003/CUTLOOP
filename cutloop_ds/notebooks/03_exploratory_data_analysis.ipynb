## Statistical & Visual Analysis of Fast-Fashion Garments & TexSortML Benchmark
This notebook visualizes brand representation, category frequency, material blend ratios, circularity index distribution, and image class splits.


```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

base_dir = os.path.abspath('..')
garments_path = os.path.join(base_dir, 'datasets', 'processed', 'garments_cleaned.csv.gz')
manifest_path = os.path.join(base_dir, 'datasets', 'processed', 'image_splits_manifest.csv')

df_g = pd.read_csv(garments_path)
df_img = pd.read_csv(manifest_path)
print('Garments:', df_g.shape, '| Images:', df_img.shape)
```

### 1. Brand & Category Breakdown


```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
df_g['brand'].value_counts().plot(kind='bar', ax=axes[0], color=['#1b4332', '#40916c'], title='Garments by Brand')
df_g['parent_category'].value_counts().head(10).plot(kind='barh', ax=axes[1], color='#2d6a4f', title='Top 10 Categories')
plt.tight_layout()
plt.show()
```

### 2. Recyclability & Material Composition


```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
df_g['dominant_fiber'].value_counts().head(8).plot(kind='bar', ax=axes[0], color=sns.color_palette('mako', 8), title='Dominant Fiber Frequency')
sns.histplot(df_g['circularity_score'], bins=20, kde=True, ax=axes[1], color='#2d6a4f')
axes[1].set_title('Circularity Index Distribution')
plt.tight_layout()
plt.show()
```

### 3. TexSortML Image Benchmark Split Breakdown


```python
pd.crosstab(df_img['category'], df_img['split']).plot(kind='bar', stacked=True, figsize=(12, 5), color=['#1b4332', '#52b788', '#b7e4c7'], title='TexSortML Splits')
plt.show()
```
